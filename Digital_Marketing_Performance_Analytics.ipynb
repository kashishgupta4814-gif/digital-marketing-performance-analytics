```python
overall_romi = (
    (total_revenue - total_marketing_spend)
    / total_marketing_spend
) * 100

print("Overall ROMI:", round(overall_romi, 2), "%")
```

```python
romi_analysis = df.groupby('campaign_name').agg({
    'mark_spent': 'sum',
    'revenue': 'sum'
})

romi_analysis
```

```python
romi_analysis['ROMI (%)'] = (
    (romi_analysis['revenue'] - romi_analysis['mark_spent'])
    / romi_analysis['mark_spent']
) * 100
```

```python
romi_analysis.round(2)
```

```python
romi_ranking = romi_analysis.sort_values(
    'ROMI (%)',
    ascending=False
)

romi_ranking.round(2)
```

```python
best_romi_campaign = romi_analysis['ROMI (%)'].idxmax()

best_romi_value = romi_analysis['ROMI (%)'].max()

print("Highest ROMI Campaign:", best_romi_campaign)
print("ROMI:", round(best_romi_value, 2), "%")
```

```python
import pandas as pd
df= pd.read_csv("marketing_spending.csv")
df.head()
```

```python
df[['mark_spent', 'revenue']].head()
```

```python
df[['mark_spent', 'revenue']].isnull().sum()
```

```python
X = df[['mark_spent']]
y = df['revenue']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
coefficient = model.coef_[0]

print("Regression Coefficient:", coefficient)
```

```python
intercept = model.intercept_

print("Intercept:", intercept)
```

```python
print(
    f"Revenue = {intercept:.2f} + "
    f"({coefficient:.2f} × Marketing Spend)"
)
```

```python
y_pred = model.predict(X_test)
```

```python
comparison = pd.DataFrame({
    'Actual Revenue': y_test.values,
    'Predicted Revenue': y_pred
})

comparison.head(10)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R² Score:", round(r2, 4))
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("Mean Absolute Error:", round(mae, 2))
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("RMSE:", round(rmse, 2))
```

```python
regression_summary = pd.DataFrame({
    'Metric': [
        'R² Score',
        'Mean Absolute Error (MAE)',
        'Root Mean Squared Error (RMSE)'
    ],
    'Value': [
        round(r2, 4),
        round(mae, 2),
        round(rmse, 2)
    ]
})

regression_summary
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
total_spend = df['mark_spent'].sum()
total_revenue = df['revenue'].sum()
total_leads = df['leads'].sum()
total_orders = df['orders'].sum()
total_clicks = df['clicks'].sum()
total_impressions = df['impressions'].sum()

overall_ctr = (total_clicks / total_impressions) * 100
overall_conversion = (total_orders / total_impressions) * 100
overall_roas = total_revenue / total_spend
overall_romi = ((total_revenue - total_spend) / total_spend) * 100

print("Total Marketing Spend:", round(total_spend, 2))
print("Total Revenue:", round(total_revenue, 2))
print("Total Leads:", total_leads)
print("Total Orders:", total_orders)
print("CTR:", round(overall_ctr, 2), "%")
print("Overall Conversion:", round(overall_conversion, 2), "%")
print("ROAS:", round(overall_roas, 2))
print("ROMI:", round(overall_romi, 2), "%")
```

```python
kpi_summary = pd.DataFrame({
    'KPI': [
        'Marketing Spend',
        'Revenue',
        'Clicks',
        'Leads',
        'Orders',
        'CTR (%)',
        'Overall Conversion (%)',
        'ROAS',
        'ROMI (%)'
    ],
    'Value': [
        round(total_spend, 2),
        round(total_revenue, 2),
        total_clicks,
        total_leads,
        total_orders,
        round(overall_ctr, 2),
        round(overall_conversion, 2),
        round(overall_roas, 2),
        round(overall_romi, 2)
    ]
})

kpi_summary
```

```python
dashboard_campaign = df.groupby('campaign_name').agg({
    'mark_spent': 'sum',
    'impressions': 'sum',
    'clicks': 'sum',
    'leads': 'sum',
    'orders': 'sum',
    'revenue': 'sum'
})

dashboard_campaign['CTR (%)'] = (
    dashboard_campaign['clicks'] /
    dashboard_campaign['impressions']
) * 100

dashboard_campaign['CPL'] = (
    dashboard_campaign['mark_spent'] /
    dashboard_campaign['leads']
)

dashboard_campaign['CPO'] = (
    dashboard_campaign['mark_spent'] /
    dashboard_campaign['orders']
)

dashboard_campaign['ROAS'] = (
    dashboard_campaign['revenue'] /
    dashboard_campaign['mark_spent']
)

dashboard_campaign['ROMI (%)'] = (
    (dashboard_campaign['revenue'] -
     dashboard_campaign['mark_spent'])
    / dashboard_campaign['mark_spent']
) * 100

dashboard_campaign.round(2)
```

```python
best_campaign = dashboard_campaign['ROMI (%)'].idxmax()
best_campaign_romi = dashboard_campaign['ROMI (%)'].max()

print("Best Campaign:", best_campaign)
print("ROMI:", round(best_campaign_romi, 2), "%")
```

```python
top_revenue_campaign = dashboard_campaign['revenue'].idxmax()

print("Highest Revenue Campaign:", top_revenue_campaign)
print(
    "Revenue:",
    round(
        dashboard_campaign.loc[
            top_revenue_campaign,
            'revenue'
        ],
        2
    )
)
```
